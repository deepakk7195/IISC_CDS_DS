```python
import os

def find_files(directory):
    """
    Finds all files in a directory, including those in subdirectories.

    Args:
        directory (str): The path to the directory to search.

    Returns:
        list: A list of file paths.
    """
    file_paths = []
    for root, _, files in os.walk(directory):
        for file in files:
            if '.txt' in file:
                file_path = os.path.join(root, file)
                file_paths.append(file_path)
    return file_paths

# Example usage:
directory_path = "C:/Users/Deepak Harishchandra/Downloads/"  # Replace with the actual path
files = find_files(directory_path)
for file in files:
    print(file)
```
