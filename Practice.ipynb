```python
def sum_diag(matrix):
    if not matrix:
        return 0
    main_diag_sum = sum(matrix[i][i] for i in range(len(matrix)))
    cross_diag_sum = sum(matrix[i][len(matrix)-i-1] for i in range(len(matrix)))
    return main_diag_sum,cross_diag_sum
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]  
print("List Matrix - ",matrix)
arr = np.array(matrix)
print("Numpy Array - ",arr)
main_diagonal_sum, other_diagonal_sum = sum_diag(matrix)
print(f"Sum of the main diagonal: {main_diagonal_sum}")
print(f"Sum of the other diagonal: {other_diagonal_sum}")
```

```python
# Selection Sort (Bringing smalest to the start / end of the array)- 
def selectionsort(arr):
    n = len(arr)
    for i in range(n):
        min_ind = i
        for j in range(i+1,n):
            if arr[j] < arr[min_ind]:
                min_ind = j
        arr[i], arr[min_ind] = arr[min_ind], arr[i]
    return arr
  
arr = [5,3,3,7,8,4,2,1,2,0,9,1]            
print(selectionsort(arr))           
```

```python
# Bubble Sort ( swapping adjacents if in wrong order) - 

def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n-i-1):
            if arr[j] > arr[j+1]:
                arr[j], arr[j+1] = arr[j+1], arr[j]
                swapped = True
        if not swapped:
            break
    return arr
nums = [5, 1, 4, 2, 8]
sorted_nums = bubble_sort(nums)
print("Sorted array:", sorted_nums)
```

```python
# Merge Sort - 

def merge_sort(arr):
    if len(arr) <= 1:
        return arr

    # Step 1: Divide
    mid = len(arr) // 2
    left_half = merge_sort(arr[:mid])
    right_half = merge_sort(arr[mid:])

    # Step 2: Merge sorted halves
    return merge(left_half, right_half)

def merge(left, right):
    result = []
    i = j = 0

    # Merge elements in sorted order
    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # Add any remaining elements
    result.extend(left[i:])
    result.extend(right[j:])

    return result

# Example usage
nums = [38, 27, 43, 3, 9, 82, 10]
sorted_nums = merge_sort(nums)
print("Sorted array:", sorted_nums)
```

```python
# Quick Sort - Pivoting around a center. Smaller to left, larger to right

def quicksort(arr):
    if len(arr) <= 1:
        return arr
    pivot  = arr[-1]
    left = [x for x in arr[:-1] if x <= pivot]
    right = [x for x in arr[:-1] if x > pivot]
    return quicksort(left) + [pivot] + quicksort(right)

nums = [10, 7, 8, 9, 1, 5]
sorted_nums = quicksort(nums)
print("Sorted array:", sorted_nums)    
```

```python
# Funciton to flatten a nested list - 
def flatten(lst):
    result = []
    for i in lst:
        if isinstance(i, list):
            result.extend(flatten(i))
        else:   
            result.append(i)
    return result

nested_list = [1, [2, 3, [4, 5]], 6]
print(flatten(nested_list))  # Output: [1, 2, 3, 4, 5, 6]
```

```python
# 2nd largest element in array w/o sorting - 

def second_highest(arr):
    if len(arr) < 2:
        return None  
    highest = arr[0]
    second = None  
    for num in arr:
        if num > highest: 
            second = highest
            highest = num
        elif second is None or (num > second and num != highest): 
            second = num
    return second
arr = [5, 1, 2, 9, 7, 9]
print(second_highest(arr))  
```

```python
# CHECK IF AN ARRAY IS SORTED ASC using Recursion - 

def issorted(i, ary):
    if i == len(ary) - 1:
        return True
    if ary[i] > ary[i+1]:
        return False
    return issorted(i+1, ary)
arr = [1,2,3,4,5,2]
print(issorted(0,arr))
```

```python
 # remove duplicates from an array in O(n) - 
def remove_duplicates(arr):
    seen = {}
    result = []
    for num in arr:
        if num not in seen:
            seen[num] = True
            result.append(num)
    return result

arr = [1, 3, 2, 2, 1, 5, 3, 4]
print(remove_duplicates(arr))  # Output: [1, 3, 2, 5, 4]    
```

```python
# Given two sorted arrays nums1 and nums2, return an array that contains the union of these two arrays. The elements in the union must be in 
#ascending order. The union of two arrays is an array where all values are distinct and are present in either the first array, the second array, 
#or both.

def unioneee(arr1, arr2):
    arr = arr1 + arr2
    dist = []
    for i in arr:
        if i not in dist:
            dist.append(i)
    for i in range(len(dist)-1):
        if dist[i] > dist[i+1]:
            dist[i], dist[i+1] = dist[i+1], dist[i]
    return dist        
nums1 = [1, 2, 3, 4, 5]
nums2 = [1, 2, 7]
print(unioneee(nums1, nums2))
```

```python
# Child list in a parent list. Ordered or just all elements are present
#1. Orderly / Contiguous manner - 
def is_sublist(sublist, parent_list):
    for i in range(len(parent_list) - len(sublist) + 1):
        if parent_list[i:i + len(sublist)] == sublist: # slicing list & checking if the slice is sublist
            return True
    return False

# Plain check - 
def is_sublist(sublist, parent_list):
    return all(item in sublist for item in parent_list)

#Converting in Set - 
def is_sublist(sublist, parent_list):
    return set(sublist).issubset(set(parent_list))

# Converting in String - 
def is_sublist(sublist, parent_list):
    sublist_str = ''.join(map(str, sublist))
    parent_list_str = ''.join(map(str, parent_list))
    return sublist_str in parent_list_str
```

```python
# Write a function that returns the sum of two numbers represented as strings
def add_strings(num1, num2):
    res, carry, i, j = "", 0, len(num1) - 1, len(num2) - 1
    while i >= 0 or j >= 0 or carry:
        n1 = int(num1[i]) if i >= 0 else 0
        n2 = int(num2[j]) if j >= 0 else 0
        temp_sum = n1 + n2 + carry
        res = str(temp_sum % 10) + res
        carry = temp_sum // 10
        i, j = i - 1, j - 1
    return res
n1 = "123"
n2 = "789"
print(add_strings(n1, n2))
```

```python
# Returning grams out of ngrams - 
tokens = ["my","name","is","deepak","khadke",".","I","am","from","jalgaon",""]
def ngrams(tokens,n):
    l = len(tokens)
    grams = [tokens[i:i+n] for i in range(l-n+1)]
    return grams
print(ngrams(tokens,3))
# Using generator instead of the list - 
def gen_ngrams(tokens,n):
    l = len(tokens)
    for i in range(l-n+1):
        yield tokens[i:i+n]
ngrams_gen = gen_ngrams(tokens,3)
print(ngrams_gen)
print([i for i in ngrams_gen])
print(type(ngrams_gen))
```

```python
# Find index of 1st non repeating integer in the array
# Can use dictionary to keep track of freq of each integer. Then iterate in array to find the 1st.
def first_nonrepeating_int(arr):
    freq_dict = {} # num & it's freq
    # Count freq of each integer
    for index, num in enumerate(arr):
        if num in freq_dict:
            freq_dict[num]+=1
        else:
            freq_dict[num]=1
    #Fetching the 1st - 
    for index, num in enumerate(arr):
        if freq_dict[num]==1:
            return index
    return -1
```

```python
#Question 18: Implement a binary search algorithm
def binary_search(arr, target):
    left, right = 0 , len(arr) - 1
    while left <= right:
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid
        elif arr[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
a = [1,2,3,4,5,6]
binary_search(a,3)
```

```python
# Example dataset (replace this with your data)
data = pd.DataFrame({
    'Characteristic': ['Age', 'Income', 'Credit Score'],
    'Baseline': [30, 40, 700],
    'Current': [32, 42, 680]
})

# Define the number of bins for histograms
num_bins = 10

# Function to calculate PSI for a characteristic
def calculate_psi(baseline, current):
    # Create histograms for both populations
    baseline_hist, _ = np.histogram(baseline, bins=num_bins)
    current_hist, _ = np.histogram(current, bins=num_bins)
    # Calculate the percentage of each population in each bin
    p_baseline = baseline_hist / len(baseline)
    p_current = current_hist / len(current)
    
    # Handle cases where p_baseline or p_current is zero
    p_baseline = np.where(p_baseline == 0, 1e-10, p_baseline)
    p_current = np.where(p_current == 0, 1e-10, p_current)
    
    # Calculate the PSI value for the characteristic
    psi = np.sum((p_current - p_baseline) * np.log(p_current / p_baseline))
    
    return psi
# Calculate PSI for each characteristic
psi_values = []
for row in data.itertuples(index=False):
    psi = calculate_psi(list(row.Baseline), list(row.Current))
    psi_values.append(psi)

# Add PSI values to the dataset
data['PSI'] = psi_values

# Display the results
print("Characteristic Analysis Report:")
print(data[['Characteristic', 'PSI']])           
```

#### Find continuos subarray(min 1 number) having largest sum & return it

- Brute Force => just compute every single subarray from start to end. Repeat for every number following on. But time complexty very very high
- Solution - 
- prev lesser value doesn't matter. Thus ignore. Calculate max sum upto a number where the number is decreasing the max sum value. Calculate such max values and return.

```python
nums = [2,3,4,5,6]
def contiguoussubarray(nums):
    maxsub = nums[0]
    cursum = 0
    for n in nums:
        if cursum < 0:
            cursum = 0
        cursum+=n
        maxsub=max(maxsub,cursum)
    return maxsub          
```

```python
prod = contiguoussubarray(nums)
prod
```

### Maximum Product subarray - Dynamic Programming

```python
nums = [2,3,4,5,6,-6,-4,0,-4,5,6,-7,0,-5,-2,2]
def contiguoussubarrayprod(nums):
    res=max(nums) # coz when array has 0 Or -1, res becomes invalid.
    currmin,currmax=1
    for n in nums:
        if n==0:
            currmin,currmax=1
            continue
        tmp=n*currmax # to keep track of what currmax is and what currmin is
        currmax = max(n*currmax,n*currmin,n)
        currmin = min(tmp,n*currmin,n)
        res=max(res,currmax)
    return res
```

```python
#pythagorean -
def hasPythagoreanTriplet(nums):
    squared_nums = set([num * num for num in nums])  # Create a set of squared values
    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            # Calculate the sum of squares of two numbers
            sum_of_squares = nums[i] * nums[i] + nums[j] * nums[j]
            if sum_of_squares in squared_nums:
                return True            
    return False
```

```python
#the longest consecutive subsequence in a list of numbers.
def longest_consecutive_subsequence(nums):
    if not nums:
        return []
    nums = sorted(set(nums))
    longest_streak = []
    current_streak = [nums[0]]
    for i in range(1, len(nums)):
        if nums[i] - nums[i - 1] == 1:
            current_streak.append(nums[i])
        else:
            if len(current_streak) > len(longest_streak):
                longest_streak = current_streak
            current_streak = [nums[i]]

    return longest_streak if len(longest_streak) > len(current_streak) else current_streak

numbers = [1, 2, 3, 5, 6, 7, 8, 10]
print(longest_consecutive_subsequence(numbers))
```

```python
# Encapsulation - Bundling all Variables & methods into single unit (class)

class Car:
    def __init__(self, make, model):
        self._make = make  # Encapsulated attribute
        self._model = model  # Encapsulated attribute

    def get_make(self):  # Getter method
        return self._make

    def get_model(self):  # Getter method
        return self._model

    def set_make(self, make):  # Setter method
        self._make = make

    def set_model(self, model):  # Setter method
        self._model = model
```

```python
# Abstraction - Hiding complex implementation details & only expose utilizable layer

from abc import ABC, abstractmethod

class Shape(ABC):  # Abstract class
    @abstractmethod
    def area(self):
        pass

class Circle(Shape):
    def __init__(self,radius):
        self.radius = radius
    def area(self):
        return 3.14 * self.radius **2

class Rectangle(Shape):
    def __init__(self,length,width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width   
```

```python
circle = Circle(5)
print(circle.area())  # Output: 78.5
rectangle = Rectangle(4, 6)
print(rectangle.area())  # Output: 24
```

```python
# Shallow Copy - Copies reference to the nested object, not the actual data. Hence, both are changed
import copy

list1 = [[1, 2, 3], [4, 5, 6]]
list2 = copy.copy(list1)  # Shallow Copy

list2[0][0] = 100  # Modifying nested list

print(list1)  # Output: [[100, 2, 3], [4, 5, 6]]
print(list2)
```

```python
# Deep Copy = Creates a completely independent copy, hence changes in the copy don't replicate in Original
list1 = [[1, 2, 3], [4, 5, 6]]
list2 = copy.deepcopy(list1)  # Deep Copy

list2[0][0] = 100  # Modifying nested list

print(list1)  # Output: [[1, 2, 3], [4, 5, 6]]
print(list2)
```
