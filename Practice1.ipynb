```python
import numpy as np
import random
import matplotlib.pyplot as plt
```

```python
lst = []
for i in range(1,10001):
    out = np.random.randint(0,2,i)
    prob = np.sum(out)/i
    lst.append(prob)
```

```python
lst[:10]
```

```python
plt.figure(figsize = (20,5))
plt.plot(lst)
```
